Supervised Learning

Supervised Learning is when the dataset already contains labels/answers.
The model learns by looking at input–output pairs and tries to predict the correct output for new data.
Example: You give the model pictures (input) and their correct categories (labels).

Unsupervised Learning

Unsupervised Learning is when the dataset has no labels.
The model explores the data and tries to find patterns, clusters, or structure on its own.
Example: Grouping similar customers based on behavior without knowing categories beforehand.

Reinforcement Learning

Reinforcement Learning is learning through rewards and penalties.
An agent interacts with an environment, takes actions, and improves by maximizing rewards.
Example: Training a robot or game agent to learn the best strategy through trial and error.
